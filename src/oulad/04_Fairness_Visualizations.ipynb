```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# 1. Set styles for a clean, cohesive layout matching your friend's theme
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 11, 
    'axes.labelsize': 12, 
    'axes.titlesize': 13,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11
})

# 2. Point directly to your results directory
RESULTS_DIR = Path("./results/oulad")
csv_path = RESULTS_DIR / "oulad_mitigation_results.csv"

# Check if the file exists before running
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find {csv_path}. Make sure notebook 3 ran completely first!")

# Load the saved data instantly!
df = pd.read_csv(csv_path)

# 3. Create a 2-row multi-panel figure layout
# Row 1: Key Performance Metrics | Row 2: Fairness Disparity Metrics
fig, axes = plt.subplots(2, 3, figsize=(18, 11))
fig.suptitle("OULAD Model Performance vs. Fairness Mitigation Trade-offs", fontsize=16, fontweight='bold', y=0.98)

# Group setups matching your data schema
perf_metrics = ['accuracy', 'f1']
fair_metrics = ['Demographic Parity', 'Equal Opportunity', 'Equalized Odds']

# --- ROW 1: PERFORMANCE METRICS (Accuracy & F1) ---
for i, metric in enumerate(perf_metrics):
    sns.barplot(
        data=df, x='sensitive_attr', y=metric, hue='miti_method', 
        ax=axes[0, i], palette="Set2", edgecolor="0.2"
    )
    axes[0, i].set_title(f"Overall Model {metric.capitalize()}", fontweight='bold')
    axes[0, i].set_ylabel(metric.capitalize())
    axes[0, i].set_xlabel("")
    axes[0, i].set_ylim(0, 1.0) # Normalizes performance scaling
    if axes[0, i].get_legend():
        axes[0, i].get_legend().remove()

# Turn off the empty 3rd slot on the top row gracefully
axes[0, 2].axis('off')

# --- ROW 2: FAIRNESS DISPARITY METRICS ---
for i, metric in enumerate(fair_metrics):
    sns.barplot(
        data=df, x='sensitive_attr', y=metric, hue='miti_method', 
        ax=axes[1, i], palette="Set2", edgecolor="0.2"
    )
    axes[1, i].set_title(f"{metric} Difference ↓", fontweight='bold')
    axes[1, i].set_ylabel("Disparity (Lower is Fairer)")
    axes[1, i].set_xlabel("Sensitive Demographic Attribute")
    if axes[1, i].get_legend():
        axes[1, i].get_legend().remove()

# Format the X-axis labels to look perfectly clean (e.g., 'Age Band', 'Region')
for ax in axes[1]:
    ax.set_xticklabels([label.get_text().replace('_', ' ').title() for label in ax.get_xticklabels()])

# 4. Extract legend handles and position them cleanly in the top right blank gap
handles, labels = axes[1, 0].get_legend_handles_labels()
axes[0, 2].legend(handles, labels, loc='center', title="Mitigation Methods", frameon=True, shadow=True, fontsize=12)

plt.tight_layout()

# 5. Save the final image directly into your oulad folder for GitLab deployment
plot_path = RESULTS_DIR / "oulad_performance_fairness_tradeoffs.png"
plt.savefig(plot_path, dpi=300, bbox_inches='tight')
plt.show()

print(f"🎉 Success! Your chart has been generated and saved to: {plot_path}")
```

```python
# Paste this into your 4th notebook to create a Pareto Frontier Plot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

df = pd.read_csv("./results/oulad/oulad_mitigation_results.csv")

plt.figure(figsize=(10, 6))
# Plotting F1-score vs Demographic Parity for Gender
gender_df = df[df['sensitive_attr'] == 'gender']

sns.scatterplot(
    data=gender_df, 
    x='Demographic Parity', 
    y='f1', 
    hue='miti_method', 
    style='model', 
    s=150, 
    edgecolor='black',
    palette='Set2'
)

plt.title("Pareto Frontier: F1-Score vs. Demographic Parity (Gender Attribute)", fontweight='bold', fontsize=14)
plt.xlabel("Demographic Parity Disparity ↓ (Lower is Fairer)")
plt.ylabel("Model F1-Score ↑ (Higher is Better)")
plt.axvline(x=0.05, color='red', linestyle='--', alpha=0.6, label='Ideal Fairness Bound (<0.05)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', shadow=True)
plt.tight_layout()
plt.savefig("./results/oulad/pareto_frontier_gender.png", dpi=300)
plt.show()
```

```python
# Code to generate a Delta Impact Heatmap
pivot_df = df.pivot_table(index=['model', 'sensitive_attr'], columns='miti_method', values='Demographic Parity')
# Calculate reduction in bias from Baseline
delta_heatmap = pivot_df.sub(pivot_df['Baseline'], axis=0).drop(columns=['Baseline'])

plt.figure(figsize=(11, 6))
sns.heatmap(delta_heatmap, annot=True, cmap="YlGnBu", fmt=".4f", linewidths=0.5, cbar_kws={'label': 'Disparity Reduction (Negative implies fairer)'})
plt.title("Algorithmic Shift: Net Change in Demographic Parity Relative to Baseline", fontweight='bold', fontsize=13)
plt.tight_layout()
plt.savefig("./results/oulad/mitigation_delta_heatmap.png", dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load data
df = pd.read_csv("./results/oulad/oulad_mitigation_results.csv")

# ==========================================
# PLOT 1: THE CLEANED MULTI-PANEL GRID
# ==========================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle("OULAD Algorithmic Auditing: Performance vs. Fairness Disparity", fontsize=16, fontweight='bold', y=0.98)

perf_metrics = ['accuracy', 'f1']
fair_metrics = ['Demographic Parity', 'Equal Opportunity', 'Equalized Odds']
clean_labels = ['Gender', 'Region', 'Age Band']

# Row 1: Performance
for i, metric in enumerate(perf_metrics):
    sns.barplot(data=df, x='sensitive_attr', y=metric, hue='miti_method', ax=axes[0, i], palette="Set2", edgecolor="0.2")
    axes[0, i].set_title(f"Overall Model {metric.capitalize()}", fontweight='bold', pad=10)
    axes[0, i].set_ylim(0, 1.0)
    axes[0, i].get_legend().remove()
    axes[0, i].set_xlabel("")

axes[0, 2].axis('off') # Hide empty slot for legend placeholder

# Row 2: Fairness
for i, metric in enumerate(fair_metrics):
    sns.barplot(data=df, x='sensitive_attr', y=metric, hue='miti_method', ax=axes[1, i], palette="Set2", edgecolor="0.2")
    axes[1, i].set_title(f"{metric} Disparity ↓", fontweight='bold', pad=10)
    axes[1, i].set_ylabel("Disparity (Lower is Fairer)")
    axes[1, i].set_xlabel("Demographic Attribute")
    axes[1, i].set_xticks(range(len(clean_labels)))
    axes[1, i].set_xticklabels(clean_labels)
    axes[1, i].get_legend().remove()

# Place a clean, elegant legend inside that empty top-right panel
handles, labels = axes[1, 0].get_legend_handles_labels()
axes[0, 2].legend(handles, labels, loc='center', title="Mitigation Strategies", frameon=True, shadow=True, fontsize=11)

plt.subplots_adjust(hspace=0.4, wspace=0.3) # Crucial: Fixes the overlapping labels!
plt.savefig("./results/oulad/oulad_clean_panel_grid.png", dpi=300, bbox_inches='tight')
plt.show()

# ==========================================
# PLOT 2: THE ENHANCED PARETO FRONTIER
# ==========================================
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, x='Demographic Parity', y='f1', hue='miti_method', style='model', 
    s=200, edgecolor='black', palette='Set2', alpha=0.9
)
plt.title("OULAD Pareto Frontier: Algorithmic Trade-Offs (F1-Score vs. Fairness Bias)", fontweight='bold', fontsize=13)
plt.xlabel("Demographic Parity Disparity ↓ (Lower is Fairer)")
plt.ylabel("Model F1-Score ↑ (Higher is Better)")
plt.axvline(x=0.05, color='gray', linestyle='--', alpha=0.5, label='Fairness Target (<0.05)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True, shadow=True)
plt.tight_layout()
plt.savefig("./results/oulad/oulad_clean_pareto.png", dpi=300, bbox_inches='tight')
plt.show()
```
